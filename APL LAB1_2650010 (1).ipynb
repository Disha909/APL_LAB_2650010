```python
import math
import matplotlib.pyplot as plt
import random

ITER = 0
sizes_arr = [10, 25, 50, 100, 250, 500, 1000]
rand_arr = [random.randint(0, i) for i in range(1000)]
print(rand_arr)
```

```python
# LINEAR SEARCH
def linear_search(arr, target):
    global ITER
    for i in range(len(arr)):
        ITER += 1
        if arr[i] == target:
            return i
    return -1
```

```python
# BINARY SEARCH
def binary_search(arr, target):
    global ITER
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            ITER += 1
            low = mid + 1
        else:
            ITER += 1
            high = mid - 1

    return -1
```

```python
ITER = 0
res = linear_search(rand_arr[:100], rand_arr[100])
print(rand_arr[:100])
print(res)
print(ITER)
```

```python
ITER = 0
res = binary_search(sorted(rand_arr[:100]), sorted(rand_arr)[100])
print(sorted(rand_arr[:100]))
print(res)
print(ITER)
```

```python
lin_iter_best = []
for i in sizes_arr:
    ITER = 0
    input_arr = sorted(rand_arr[:i])
    input_query = input_arr[0]
    linear_search(input_arr, input_query)
    lin_iter_best.append(ITER)

print(lin_iter_best)
```

```python
lin_iter_worst = []
for i in sizes_arr:
    ITER = 0
    input_arr = sorted(rand_arr[:i])
    input_query = input_arr[i-1]
    linear_search(input_arr, input_query)
    lin_iter_worst.append(ITER)

print(lin_iter_worst)
```

```python
lin_iter_avg = []
for i in sizes_arr:
    ITER = 0
    input_arr = sorted(rand_arr[:i])
    input_query = input_arr[int(i//2)]
    linear_search(input_arr, input_query)
    lin_iter_avg.append(ITER)

print(lin_iter_avg)
```

```python
bin_iter_best = []
for i in sizes_arr:
    ITER = 0
    input_arr = sorted(rand_arr[:i])
    input_query = input_arr[int(i//2)]
    # print(input_query, input_arr)
    binary_search(input_arr, input_query)
    bin_iter_best.append(ITER)

print(bin_iter_best)
```

```python
bin_iter_worst = []
for i in sizes_arr:
    ITER = 0
    input_arr = sorted(rand_arr[:i])
    input_query = sorted(input_arr)[i-1]
    binary_search(input_arr, input_query)
    bin_iter_worst.append(ITER)

print(bin_iter_worst)
```

```python
bin_iter_avg = []
for i in sizes_arr:
    ITER = 0
    input_arr = sorted(rand_arr[:i])
    input_query = input_arr[random.randint(0,i-1)]
    binary_search(input_arr, input_query)
    bin_iter_avg.append(ITER)

print(bin_iter_avg)
```

```python
plt.plot(sizes_arr, lin_iter_best, label="Best Case")
plt.plot(sizes_arr, lin_iter_avg, label="Avg Case")
plt.plot(sizes_arr, lin_iter_worst, label="Worst Case")

plt.xlim(left=0)

plt.legend()
plt.xlabel("Input Size")
plt.ylabel("No. of Comparisons")
plt.title("Linear Search")
plt.grid(visible=True, linestyle=':')
plt.show()
```

```python
plt.plot(sizes_arr, bin_iter_best, label="Best Case")
plt.plot(sizes_arr, bin_iter_avg, label="Avg Case")
plt.plot(sizes_arr, bin_iter_worst, label="Worst Case")

plt.xlim(left=0)

plt.legend()
plt.xlabel("Input Size")
plt.ylabel("No. of Comparisons")
plt.title("Binary Search")
plt.grid(visible=True, linestyle=':')
plt.show()
```
